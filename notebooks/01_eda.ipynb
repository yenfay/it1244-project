```python
#adding headers to txt 
import os 
import pandas as pd

raw = '../data/raw'
processed = '../data/processed'
os.makedirs(processed, exist_ok=True)

col_names = (['time'] 
            + [f"L{i}" for i in range(1, 9)]    
            + [f"R{i}" for i in range(1, 9)] 
            + ["L_total", "R_total"])

for file in os.listdir(raw):
    if not file.endswith('.txt'):
        continue
    else:
        df = pd.read_csv(os.path.join(raw, file), sep=r'\s+', header =None, names= col_names) #builds data/raw/GaCo01_01.txt
        df.to_csv(os.path.join(processed, file.replace(".txt", ".csv")), index=False)

```
